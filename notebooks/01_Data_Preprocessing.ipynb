```python
# ==========================================
# 1. Import Libraries
# ==========================================

import pandas as pd
import numpy as np
import ast
import os

# ==========================================
# 2. Load Datasets
# ==========================================

movies = pd.read_csv("../dataset/movies.csv")
credits = pd.read_csv("../dataset/credits.csv")
keywords = pd.read_csv("../dataset/keywords.csv")

print("Movies :", movies.shape)
print("Credits:", credits.shape)
print("Keywords:", keywords.shape)

# ==========================================
# 3. Convert ID Columns to Integer
# ==========================================

movies["id"] = pd.to_numeric(movies["id"], errors="coerce")
credits["id"] = pd.to_numeric(credits["id"], errors="coerce")
keywords["id"] = pd.to_numeric(keywords["id"], errors="coerce")

movies = movies.dropna(subset=["id"])
credits = credits.dropna(subset=["id"])
keywords = keywords.dropna(subset=["id"])

movies["id"] = movies["id"].astype(int)
credits["id"] = credits["id"].astype(int)
keywords["id"] = keywords["id"].astype(int)

# ==========================================
# 4. Remove Duplicate Movie IDs
# ==========================================

movies = movies.drop_duplicates(subset="id", keep="first")
credits = credits.drop_duplicates(subset="id", keep="first")
keywords = keywords.drop_duplicates(subset="id", keep="first")

print("\nDuplicate IDs Remaining")
print("-----------------------")
print("Movies   :", movies["id"].duplicated().sum())
print("Credits  :", credits["id"].duplicated().sum())
print("Keywords :", keywords["id"].duplicated().sum())

# ==========================================
# 5. Merge Datasets
# ==========================================

ott_df = movies.merge(credits, on="id", how="left")
ott_df = ott_df.merge(keywords, on="id", how="left")

print("\nMerged Dataset Shape:")
print(ott_df.shape)

# ==========================================
# 6. Create Processed Folder (if it doesn't exist)
# ==========================================

os.makedirs("../dataset/processed", exist_ok=True)

# ==========================================
# 7. Save Merged Dataset
# ==========================================

ott_df.to_csv("../dataset/processed/merged_dataset.csv", index=False)

print("\nMerged dataset saved successfully!")
```

```python
import pandas as pd

# Convert budget to numeric
ott_df["budget"] = pd.to_numeric(ott_df["budget"], errors="coerce")

# Check available financial data
print("Movies with Budget > 0 :", (ott_df["budget"] > 0).sum())
print("Movies with Revenue > 0 :", (ott_df["revenue"] > 0).sum())
print("Movies with BOTH Budget & Revenue > 0 :",
      ((ott_df["budget"] > 0) & (ott_df["revenue"] > 0)).sum())
```

```python
# Convert release_date to datetime
ott_df["release_date"] = pd.to_datetime(
    ott_df["release_date"],
    errors="coerce"
)

# Check data types
print(ott_df.dtypes)
```

```python
# Create Profit
ott_df["profit"] = ott_df["revenue"] - ott_df["budget"]

# Create ROI
ott_df["roi"] = (ott_df["revenue"] - ott_df["budget"]) / ott_df["budget"]

# Display first few rows
ott_df[["budget", "revenue", "profit", "roi"]].head()
```

```python
# Keep only movies with valid budget and revenue
model_df = ott_df[
    (ott_df["budget"] > 0) &
    (ott_df["revenue"] > 0)
].copy()

# Recalculate Profit
model_df["profit"] = model_df["revenue"] - model_df["budget"]

# Recalculate ROI
model_df["roi"] = model_df["profit"] / model_df["budget"]

print(model_df.shape)

print(model_df["roi"].describe())
```

```python
model_df = final_df.copy()
```

```python
# Create Franchise feature
model_df["franchise"] = model_df["belongs_to_collection"].notna().astype(int)

# Check distribution
print(model_df["franchise"].value_counts())

# Remove original JSON column
model_df.drop(columns=["belongs_to_collection"], inplace=True)
```

```python
# Release Year
model_df["release_year"] = model_df["release_date"].dt.year

# Release Month
model_df["release_month"] = model_df["release_date"].dt.month

# Release Quarter
model_df["release_quarter"] = model_df["release_date"].dt.quarter

# Function to determine season
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    elif month in [9, 10, 11]:
        return "Fall"
    else:
        return None

# Release Season
model_df["release_season"] = model_df["release_month"].apply(get_season)

# Remove original release_date column
model_df.drop(columns=["release_date"], inplace=True)

# Check the new columns
model_df[[
    "release_year",
    "release_month",
    "release_quarter",
    "release_season"
]].head()
```

```python
import ast

# Extract genre names
model_df["genre_list"] = model_df["genres"].apply(
    lambda x: [i["name"] for i in ast.literal_eval(x)]
)

# Check first few rows
model_df[["genres", "genre_list"]].head()
```

```python
from collections import Counter

# Count all genres
genre_counts = Counter()

for genres in model_df["genre_list"]:
    genre_counts.update(genres)

# Convert to DataFrame
genre_df = (
    pd.DataFrame(
        genre_counts.items(),
        columns=["Genre", "Count"]
    )
    .sort_values(by="Count", ascending=False)
)

genre_df
```

```python
genre_df.head(20)
```

```python
#Avg ROI by genre

# Explode the genre list so each genre gets its own row
genre_roi = model_df.explode("genre_list")

# Calculate statistics
genre_roi_summary = (
    genre_roi.groupby("genre_list")["roi"]
    .agg(
        Movie_Count="count",
        Average_ROI="mean",
        Median_ROI="median",
        ROI_STD="std"
    )
    .sort_values("Average_ROI", ascending=False)
)

genre_roi_summary
```

```python
#director extraction

import ast

def extract_director(crew):
    try:
        crew = ast.literal_eval(crew)
        for person in crew:
            if person["job"] == "Director":
                return person["name"]
    except:
        return None

model_df["director"] = model_df["crew"].apply(extract_director)

model_df[["director"]].head()
```

```python
#keywords

def extract_keywords(keywords):
    try:
        keywords = ast.literal_eval(keywords)
        return [k["name"] for k in keywords]
    except:
        return []

model_df["keyword_list"] = model_df["keywords"].apply(extract_keywords)

model_df[["keyword_list"]].head()
```

```python
#production companies

def extract_companies(companies):
    try:
        companies = ast.literal_eval(companies)
        return [c["name"] for c in companies]
    except:
        return []

model_df["company_list"] = model_df["production_companies"].apply(extract_companies)

model_df[["company_list"]].head()
```

```python
#spoken languages

def extract_languages(languages):
    try:
        languages = ast.literal_eval(languages)
        return [l["name"] for l in languages]
    except:
        return []

model_df["language_list"] = model_df["spoken_languages"].apply(extract_languages)

model_df[["language_list"]].head()
```

```python
#Production Countries
def extract_countries(countries):
    try:
        countries = ast.literal_eval(countries)
        return [c["name"] for c in countries]
    except:
        return []

model_df["country_list"] = model_df["production_countries"].apply(extract_countries)

model_df[["country_list"]].head()
```

```python
model_df.drop(columns=[
    "genres",
    "cast",
    "crew",
    "keywords",
    "production_companies",
    "production_countries",
    "spoken_languages"
], inplace=True)
```
