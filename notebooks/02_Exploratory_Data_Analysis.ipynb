```python
# ==========================================
# Budget Groups
# ==========================================

budget_bins = [
    0,
    10_000_000,
    30_000_000,
    60_000_000,
    100_000_000,
    model_df["budget"].max()
]

budget_labels = [
    "<10M",
    "10-30M",
    "30-60M",
    "60-100M",
    ">100M"
]

model_df["budget_group"] = pd.cut(
    model_df["budget"],
    bins=budget_bins,
    labels=budget_labels
)

budget_summary = (
    model_df
    .groupby("budget_group", observed=False)["roi"]
    .agg(
        Movie_Count="count",
        Average_ROI="mean",
        Median_ROI="median",
        ROI_STD="std"
    )
)

budget_summary
```

```python
# ==========================================
# ROI by Budget Group
# ==========================================

plt.figure(figsize=(10,6))

model_df.boxplot(
    column="roi",
    by="budget_group",
    grid=False
)

plt.title("ROI Distribution Across Budget Groups")
plt.suptitle("")
plt.xlabel("Budget Group")
plt.ylabel("ROI")

plt.show()
```

```python
# ==========================================
# 10. Average ROI by Budget Group
# ==========================================

budget_summary["Average_ROI"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average ROI Across Budget Groups")
plt.xlabel("Budget Group")
plt.ylabel("Average ROI")

plt.xticks(rotation=0)

plt.show()
```

```python
# ==========================================
# 11. Median ROI by Budget Group
# ==========================================

budget_summary["Median_ROI"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Median ROI Across Budget Groups")
plt.xlabel("Budget Group")
plt.ylabel("Median ROI")

plt.xticks(rotation=0)

plt.show()
```

```python
# ==========================================
# 12. ROI Risk Across Budget Groups
# ==========================================

budget_summary["ROI_STD"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("ROI Risk Across Budget Groups")
plt.xlabel("Budget Group")
plt.ylabel("ROI Standard Deviation")

plt.xticks(rotation=0)

plt.show()
```

```python
# ==========================================
# 13.1 Convert Genre Strings to Lists
# ==========================================

import ast

model_df["genre_list"] = model_df["genre_list"].apply(ast.literal_eval)

model_df["genre_list"].head()
```

```python
# ==========================================
# 13.2 Number of Movies per Genre
# ==========================================

from collections import Counter

genre_counter = Counter()

for genres in model_df["genre_list"]:
    genre_counter.update(genres)

genre_count = (
    pd.DataFrame(
        genre_counter.items(),
        columns=["Genre", "Movie_Count"]
    )
    .sort_values("Movie_Count", ascending=False)
)

genre_count
```

```python
# ==========================================
# 13.3 Genre Distribution
# ==========================================

plt.figure(figsize=(12,6))

plt.bar(
    genre_count["Genre"],
    genre_count["Movie_Count"]
)

plt.title("Number of Movies by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")

plt.xticks(rotation=45)

plt.show()
```

```python
# ==========================================
# 13.4 ROI Statistics by Genre
# ==========================================

genre_roi = []

for genre in genre_count["Genre"]:

    temp = model_df[
        model_df["genre_list"].apply(lambda x: genre in x)
    ]

    genre_roi.append([
        genre,
        len(temp),
        temp["roi"].mean(),
        temp["roi"].median(),
        temp["roi"].std()
    ])

genre_summary = pd.DataFrame(
    genre_roi,
    columns=[
        "Genre",
        "Movie_Count",
        "Average_ROI",
        "Median_ROI",
        "ROI_STD"
    ]
)

genre_summary = genre_summary.sort_values(
    "Average_ROI",
    ascending=False
)

genre_summary
```

```python
# ==========================================
# 13.5 Average ROI by Genre
# ==========================================

plt.figure(figsize=(12,6))

genre_summary.sort_values("Average_ROI").plot(
    x="Genre",
    y="Average_ROI",
    kind="barh",
    legend=False,
    figsize=(12,7)
)

plt.title("Average ROI by Genre")
plt.xlabel("Average ROI")
plt.ylabel("Genre")

plt.show()
```

```python
# ==========================================
# 13.6 ROI Risk by Genre
# ==========================================

genre_summary.sort_values("ROI_STD").plot(
    x="Genre",
    y="ROI_STD",
    kind="barh",
    legend=False,
    figsize=(12,7)
)

plt.title("ROI Standard Deviation by Genre")
plt.xlabel("ROI Standard Deviation")
plt.ylabel("Genre")

plt.show()
```

```python
# ==========================================
# 13.7A Filter genres with enough movies
# ==========================================

genre_filtered = genre_summary[
    genre_summary["Movie_Count"] >= 100
].copy()

genre_filtered
```

```python
# ==========================================
# 13.7B Coefficient of Variation
# ==========================================

genre_filtered["CV"] = (
    genre_filtered["ROI_STD"] /
    genre_filtered["Average_ROI"]
)

genre_filtered = genre_filtered.sort_values("CV")

genre_filtered[
    ["Genre",
     "Movie_Count",
     "Average_ROI",
     "Median_ROI",
     "ROI_STD",
     "CV"]
]
```

```python
# ==========================================
# 13.8 CV by Genre
# ==========================================

plt.figure(figsize=(12,7))

genre_filtered.sort_values("CV").plot(
    x="Genre",
    y="CV",
    kind="barh",
    figsize=(12,7),
    legend=False
)

plt.title("Coefficient of Variation by Genre")
plt.xlabel("CV (Lower = More Stable)")
plt.ylabel("Genre")

plt.show()
```

```python
# ==========================================
# 13.9 Risk-Return Matrix
# ==========================================

import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(
    genre_filtered["ROI_STD"],
    genre_filtered["Average_ROI"],
    s=120
)

# Label each point
for _, row in genre_filtered.iterrows():
    plt.text(
        row["ROI_STD"] + 0.05,
        row["Average_ROI"] + 0.03,
        row["Genre"],
        fontsize=9
    )

# Reference lines
plt.axhline(
    genre_filtered["Average_ROI"].mean(),
    color="red",
    linestyle="--",
    alpha=0.7,
    label="Average ROI"
)

plt.axvline(
    genre_filtered["ROI_STD"].mean(),
    color="green",
    linestyle="--",
    alpha=0.7,
    label="Average Risk"
)

plt.title("Risk-Return Matrix by Genre")
plt.xlabel("Risk (ROI Standard Deviation)")
plt.ylabel("Average ROI")

plt.legend()

plt.show()
```
